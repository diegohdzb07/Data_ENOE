```python
import pandas as pd 
from sklearn import tree
import matplotlib
import matplotlib.pyplot as plt
import openpyxl
datos = pd.read_csv("../databases/full_database.csv")
```

```python
datos.head()
```

```python
#Filtrar por ent
datos = datos[datos['ent'] == 7]
```

```python
datos = datos[["emp_ppal", "ano"]]
```

```python
datos_agrupados = datos.groupby(["ano", "emp_ppal"]).value_counts()
datos_agrupados = datos_agrupados.to_frame(name = "total").reset_index()#.drop("total", axis = 1)
datos_agrupados = datos_agrupados.pivot(index = "ano",columns= "emp_ppal",values= "total").fillna(0)
```
