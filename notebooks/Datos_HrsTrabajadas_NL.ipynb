```python
#Filtrar por ent
datos = datos[datos['ent'] == 19]
```

```python
datos = datos[["hrsocup", "ano"]]
```

```python
datos_agrupados = datos.groupby(["ano", "hrsocup"]).value_counts()
datos_agrupados = datos_agrupados.to_frame(name = "total").reset_index()#.drop("total", axis = 1)
datos_agrupados = datos_agrupados.pivot(index = "ano",columns= "hrsocup",values= "total").fillna(0)
```
